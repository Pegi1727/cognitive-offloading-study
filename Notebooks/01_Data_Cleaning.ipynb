# Data Cleaning and Validation
This notebook validates the `screening_data.csv` and ensures data integrity.

```r
library(readr)
d <- read_csv('screening_data.csv')
cat('Total rows:', nrow(d), '\n')
cat('Duplicates:', sum(duplicated(d$ID)), '\n')
cat('Rows failing validation (Total != P2+P4):', sum(d$Total != (d$P2 + d$P4), na.rm=TRUE))
```
